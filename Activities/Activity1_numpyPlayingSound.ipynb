8) Using your original audio's sampling rate, figure out how many samples you would need to define a "grain" of sound that is 100ms long. Make one grain from the beginning of the original audio file, "start".

```python
grain_samples = int(0.1 * fs)  
start = x[:grain_samples]     

Audio(start, rate=fs)
```

9) Using the audio arrays you have created above (reversed, masked, skipped, etc), create more grains of 50-100ms lengths then concatenate and repeat (i.e. continuously append) until you have two seconds worth of sound. Play your new sound back.

```python
sources = [x, reverse, masked, skipped]
target_len = int(2 * fs)
out = []
current = 0
while current < target_len:
    src = sources[np.random.randint(len(sources))]

    grain_len = np.random.randint(int(0.05 * fs), int(0.10 * fs))

    if src.size <= grain_len:
        start_i = 0
    else:
        start_i = np.random.randint(0, src.size - grain_len)

    grain = src[start_i:start_i + grain_len]
    out.append(grain)
    current += grain.size

new_sound = np.concatenate(out)[:target_len]
Audio(new_sound, rate=fs)
```
